```python
## Declaring and assigning a variable
age = 32
height = 6.3
name = "John"
is_student = True
## printing the variables
print("Age:", age)
print("Height:", height)
print("Name:", name)
print("Is Student:", is_student)
```

```python
## understanding variable types
## Python is a dynamically typed language, which means you don't need to declare the type of a variable explicitly. The type is inferred based on the value assigned to it.
age = 25
print(type(age)) 
```

```python
## Type Checking and Conversion
type(age)
print(type(age))
age_str = str(age)
print(age_str)
print(type(age_str))
```

```python
## Dynamic Typing
## Python allows you to change the type of a variable as the program executes
var = 10
print(var, type(var))

var = "Now I'm a string"
print(var, type(var))

var = 3.14
print(var, type(var))
```

```python
## Python Input
user_input = input("Enter your name: ")
print("Hello, " + user_input + "!")
print(type(user_input))

user_age = input("Enter your age: ")
print("You are " + user_age + " years old.")
print(type(user_age))
print(type(int(user_age)))
```

```python
### Simple Calculator
num1 = input("Enter the first number: ")
num2 = input("Enter the second number: ")

sum = float(num1) + float(num2)
difference = float(num1) - float(num2)
product = float(num1) * float(num2)
quotient = float(num1) / float(num2)
print("Sum:", sum)
print("Difference:", difference)
print("Product:", product)
print("Quotient:", quotient)
```
