## Dictionary<br/>

i) Collection of Key-Value Pairs  <br/>
ii) Mutable (Can be modified after creation)  <br/>
iii) Keys must be unique  <br/>
iv) Keys are used to access values  <br/>
v) Unordered collection (in older Python versions and insertion-ordered since Python 3.7  <br/>
vi) Defined using curly braces `{}` with `key: value` pairs  <br/>
vii) Used for fast data lookup and data mapping

```python
## Creating Dictionaries
empty_dict = {}
print(type(empty_dict))
set1 = {}
print(type(set1))
```

```python
empty_dict = dict()
empty_dict
```

```python
student = {"name": "Maruf","Age" : 32, "Grade": 24}
print(student)
print(type(student))
```

```python
## Single key is always used
student = {"name": "Maruf","Age" : 32, "name": 24}
print(student)
```

```python
## Accessing elements
student = {"name": "Maruf","Age" : 32, "Grade": "A"}
print(student["Grade"])
print(student['Age'])
```

```python
## Accessing using get() method
print(student.get("Grade"))
print(student.get("Age"))
print(student.get('last_name'))
print(student.get('last_name', 'Not available'))
```

```python
## Modifying Dictionary Elements
## Dictionary are mutable, so you can add, update or delete elements
print(student)
```

```python
student["Age"] = 33 ## update a value
print(student)
student["Address"] = "BD" ## added a new key and value

print(student)
```

```python
del student['Grade'] ## delete key and value pair 
print(student)
```

```python
## Common Dictionary Methods
keys = student.keys()
print(keys)
values = student.values()
print(values)
items = student.items()
print(items)
```

```python
## Shallow Copy
student_copy = student
print(student)
print(student_copy)
```

```python
student["name"] = "Soshi"
student_copy1 = student.copy()
print(student_copy1)
print(student)
```

```python
student["name"] = "Tauhid2"
print(student_copy1)
print(student)
```

```python
### Iterating over dictionaries 
### Loops to iterate over dictionaries, keys, values or items

## Iterating over keys
for keys in student.keys():
    print(keys)
```

```python
### Iterate over values 
for value in student.values():
    print(value)
```

```python
### Iterate over key value pairs
for key, value in student.items():
    print(f"{key} : {value}")
```

```python
### Nested Dictionaries 
students = {
    "student1" : {"name" : "Maruf", 'age': 28},
    "student2" : {"name" : "Muhammad", 'age' : 12}
}
print(students)
```

```python
## ACcess nested dictionaries
print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
### Iterating over nested dictionaries
for student_id, student_info in students.items():
    print(f"{student_id}: {student_info}")
    for key, value in student_info.items():
        print(f"{key} : {value}")
```

```python
## Dictionary Comprehension
squares = { x : x**2 for x in range(5)}
print(squares)
```

```python
## Conditional Dictionary Comprehension
evens = {x : x**2 for x in range (10) if x %2 == 0 }
print(evens)
```

```python
## Practical Examples 
numbers = [14,2,2,3,3,3,4,4,4,4,1,1]
frequency = {}
for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else: 
        frequency[number] = 1
print(frequency)
```

```python
## Merge 2 dictionaries into one
dict1 = { "a" : 1, "b" : 2}
dict2 = {"b" : 3, "c" : 4}
merged_dict = {**dict1, **dict2}
print(merged_dict)
```
