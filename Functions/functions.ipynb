```python
## Why functions?
num = 24
if num % 2 == 0:
    print("The number is even")
else:
    print("The Number is odd")
```

```python
def even_or_odd(num):
    """This function finds even or odd"""
    if num % 2 == 0:
        print("The number is even")
    else:
        print("The Number is odd")
```

```python
even_or_odd(23)
```

```python
### Functions with multiple parameters
def add(a,b):
    return a + b
result = add(2,4)
print(result)
```

```python
## Default Parameters 
def greet(name):
    print(f"Hello {name}")
greet("Maruf")
```

```python
def greet(name="Guest"):
    print(f"Hello {name}")
greet()
```

```python
### Variable length Arguments
### Positional and keywords arguments
def print_number(*args):
    for numbers in  args:
        print(numbers)
```

```python
print_number(1,2,3,4,5,"Maruf")
```

```python
### Keyword arguments
def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} : {value}")
```

```python
print_details(name="Maruf", age= 22,country = "Bd" )
```

```python
def print_both_details(*args,**kwargs):
    for val in args:
        print(f"Positional argument: {val}")
    for key, value in kwargs.items():
        print(f"{key} : {value}")
```

```python
print_both_details(1,2,"Maruf",name="Maruf", age= 22,country = "Bd")
```
