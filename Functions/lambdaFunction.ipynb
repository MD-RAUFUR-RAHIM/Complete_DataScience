Lambda Function
<br/>
Syntax</br>
lambda argument:expression

```python
def addition(a,b):
    return a+b
addition(2,3)
```

```python
addition = lambda a,b : a+b
type(addition)
print(addition(4,5))
```

```python
def even(num):
    if num % 2 ==0:
        return True
even(24)
```

```python
even1 = lambda num : num % 2 == 0
print(even(20))
```

```python
def addition(x,y,z):
    return x+y+z
addition(12,23,4)
```

```python
addition1 = lambda x,y,z : x+y+z
print(addition1(2,34,4))
```

```python
## map()
numbers = [1,2,3,4,5,6]
list(map(lambda x:x **2, numbers))
```
