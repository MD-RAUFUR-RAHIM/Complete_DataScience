The map() Function in python

```python
def squared(x):
    return x**2
squared(10)
```

```python
numbers = [1,2,3,4,5,6]
list(map(squared, numbers))
```

```python
### Lambda function with map
numbers = [1,2,3,4,5,6]
list(map(lambda x:x **2, numbers))
```

```python
### Map with multiple iterable
numbers1 = [1,2,3]
numbers2 = [4,5,6]

added_numbers = list(map(lambda x,y : x+y, numbers1, numbers2))
print(added_numbers)
```

```python
### map() to convert a list of string to integers
str_numbers = ['1','2','3']
int_numbers = list(map(int,str_numbers))
print(int_numbers)
```

```python
words=['apple','banana','cherry']
upper_word = list(map(str.upper, words))
print(upper_word)
```

```python
def get_name(person):
    return person['name']
people = [
    {'name': 'Maruf', 'age': 12},
    {'name': 'Ma', 'age': 1},
    {'name': 'Mar', 'age': 812},
    {'name': 'Maru', 'age': 2},
]
list(map(get_name,people))
```
