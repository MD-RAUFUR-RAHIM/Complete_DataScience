### List Comprehension
Basic Syntax <br>
[Expression for item in iterable] <br>
With Conditional Logic <br>
[Expression for item in iterable if condition] <br>

```python
### Basic List Comprehension
square = [num**2 for num in range(10)]
print(square)
```

```python
### List Comprehension with Condition
lst = []
for i in range(10):
    if i%2 == 0:
        lst.append(i)
print(lst)
```

```python
even_numbers = [num for num in range(10) if num%2==0]
print(even_numbers)
```

### Nested List Comprehension
[expression for item in iterable for item2 in iterable]

```python
### Nested List Comprehension
lst1 = [1,2,3,4]
lst2 = ['a', 'b', 'c','d']
pair = [[i,j] for i in lst1 for j in lst2]
print(pair)
```

```python
### List Comprehension with function calls
words = ["hello", "world", "python", "list", "comprehension"]
lengths = [len(word) for word in words]
print(lengths)
```

```python
names = ['tauhid','zunaira', 'maruf']
capitalized = [name.capitalize() for name in names]
print(capitalized)
```

```python
matrix = [ [i*j for j in range(3)] for i in range(3)]
print(matrix)
```

```python
result = [ [i,"Even" if i % 2 == 0 else "Odd"] for i in range(10)]
print(result)
```
