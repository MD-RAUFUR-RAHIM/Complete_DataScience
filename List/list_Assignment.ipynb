### Assignment 6: List Methods
Create a list of random numbers and sort it in ascending and descending order. Remove the duplicates from the list and print the modified list.

```python
import random
random_numbers = [random.randint(1,20) for _  in range(7)]
print(f"Random numbers: {random_numbers}")

sorted_number_ascending = sorted(random_numbers)
print(f"Sorted Numbers in Ascending order: {sorted_number_ascending}")

sorted_number_descending = sorted(random_numbers, reverse=True)
print(f"Sorted Numbers in Descending order: {sorted_number_descending}")

unique_numbers_only = list(set(random_numbers))
print(f"Unique numbers: {unique_numbers_only}")

duplicate_numbers_only = [num for num in random_numbers if random_numbers.count(num) > 1 ]
print(f"Duplicate Numbers only: {duplicate_numbers_only}")

```

### Assignment 7: Nested Lists

Create a nested list representing a 3x3 matrix and print the matrix. Access and print the element at the second row and third column.

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print("Matrix")
for row in matrix:
    print(row)
print(f"Element at the second row and third column: {matrix[1][2]}")
```

```python
nested_list = [[i * j for i in range(3)] for j in range(3)]
print(nested_list)
```
