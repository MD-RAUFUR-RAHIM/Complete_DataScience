```python
## while Loop
count = 0
while count < 5:
    print(count)
    count += 1
```

```python
count = 0 
while count%2 == 0:
    print(count)
    count += 1
```

```python
## Loop control Statements
## Break
for i in range(10):
    if i == 5:
        break;
    print(i)
```

```python
## continue
for i in range(10):
    if i%2 == 0:
        continue;
    print(i)
```

```python
## Nested Loops
for i in range(3):
    for j in range(2):
        print(f"i: {i} and j: {j}")
```

```python
## Examples- Calculate the sum of first n natural numbers
x = int(input("Enter a number: "))
result = 0
while x > 0:
    result += x
    x = x-1
print(f"The sum of first n natural numbers is: {result}")
```

```python
n = int(input("Enter a number: "))
sum = 0
count = 1
while count <= n:
    sum += count
    count = count + 1
print(f"The sum of first n natural numbers is: {sum}")
```

```python
## Prime Numbers
n = int(input("Enter a number: "))
for num in range(1, n+1):
    if num > 1:
        for i in range(2, num):
            if (num % i) == 0:
                break
        else:
            print(num)
```
