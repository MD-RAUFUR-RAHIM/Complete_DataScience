Sets are:<br/>
i) Collection of unique items<br/>
ii) Unordered<br/>
iii) Do not allow duplicate elements<br/>
iv) Used for membership tests and eliminating duplicate entries<br/>
v) Defined using curly braces {}<br/>
vi) Mutable (elements can be added or removed)<br/>

```python
## Create Set
my_set = {1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_set = set([1,2,3,6,5,4,5,6])
print(my_set)
```

```python
## Basic Sets Operation
## Adding and removing elements
my_set.add(7)
print(my_set)
my_set.add(7)
print(my_set)## as it is a duplicate element it is going to ignored
```

```python
##Remove the element from a set
my_set.remove(3)
print(my_set)
```

```python
my_set.remove(10)
```

```python
## if the element is present it would be removed
## if not present then will not give an error and continue unlike remove method
my_set.discard(11)
print(my_set) 
```

```python
## pop method
## it is First in First out FIFO
removed_element = my_set.pop()
print(removed_element)
print(my_set)
```

```python
## clear all the elements
my_set.clear()
print(my_set)
```

```python
## Set Membership Test
my_set = {1,2,3,4,5}
print(3 in my_set)
print(10 in my_set)
```

```python
## Mathematical Operation
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}
## Union
union_set = set1.union(set2)
print("Union Example",union_set)
## Intersection
intersect_set = set1.intersection(set2)
print("Intersection Example", intersect_set)

## it gives the intersection and updates the values of the first set itself
set1.intersection_update(set2)
print(set1)
```

```python
## Difference
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}
## this method will remove the common values from set1 which is 4,5,6
print(set1.difference(set2))
print(set1)
## it will update the set1 itself with the values
set1.difference_update(set2)
print("Updated set1 after difference_update method",set1)
```

```python
## Symmetric Difference. Both unique elements from the two sets with removed common elements
set1.symmetric_difference(set2)
```

```python
## Set Methods
set1 = {1,2,3,4,5,6}
set2 = {3,4,5}
## is subset
print(set1.issubset(set2))

print(set1.issuperset(set2))
```

```python
lst= [1,2,2,3,4,4,5]
set(lst)
```

```python
### Counting Unique words in text
text = "In this tutorial we are discussing about sets In In"
words = text.split() 

## convert list of words to set to get unique words

unique_words = set(words)
print(unique_words)
print(len(unique_words))
```
