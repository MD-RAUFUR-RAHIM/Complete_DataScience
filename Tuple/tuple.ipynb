Tuples are:<br/>
i) Ordered collection of items<br/>
ii) Immutable (cannot be changed after creation)<br/>
iii) Allow duplicate elements<br/>
iv) Can store different data types<br/>
v) Defined using parentheses ()<br/>
vi) Used for storing fixed collections of data<br/>

```python
### Create Tuple
empty_tuple = ()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst = list()
print(type(lst))
tpl = tuple()
print(type(tpl))
```

```python
numbers = tuple([1,2,3,4,5,6])
numbers
```

```python
list((1,2,3,4,5,6))
```

```python
mixed_tuple = (1, "Hello world!", 3.14, True)
print(mixed_tuple)
```

```python
### Tuple Methods
print(numbers.count(1))
print(numbers.index(3))
```

```python
## Packing tuple
packed_tuple = 1, "Hello",3.14
print(packed_tuple)
```

```python
## Unpacking tuple
a,b,c = packed_tuple
print(a)
print(b)
print(c)
```

```python
## Unpacking with * 
numbers = (1,2,3,4,5,6)
first, *middle, last = numbers
print(first)
print(middle)
print(last)
```

```python
## Nested List
lst=[[1,2,3,4],[6,7,8,9],[1,"Hello",2.1,4]]
lst[2][0:3]
```

```python
## Nested tuple
nested_tuple = ((1,2,3),('a','b','c'),(True, False))
nested_tuple[0]
print(nested_tuple[1][2])
```

```python
## Iterating over Nested Tuple
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item, end=" ")
    print()
```
